```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
data_path = Path("../data/car_sales.csv")
if not data_path.exists():
    raise FileNotFoundError("Add the original dataset at ../data/car_sales.csv")
df = pd.read_csv(data_path)
df.head()
```

```python
print("Shape:", df.shape)
display(df.info())
display(df.isna().sum().sort_values(ascending=False))
print("Duplicates:", df.duplicated().sum())
```

```python
clean = df.drop_duplicates().copy()
display(clean.describe(include="all"))
```

```python
clean.select_dtypes(include="number").hist(figsize=(12,8), bins=25)
plt.tight_layout()
plt.show()
```

```python
numeric = clean.select_dtypes(include="number")
if numeric.shape[1] >= 2:
    display(numeric.corr())
```
